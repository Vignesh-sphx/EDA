```python
import pandas as pd

df=pd.read_csv("C:/Users/91754/Downloads/transactions.csv")
print(df.head())
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
high_amount = df[df["Amount (INR)"] > 5000]
print(high_amount)
```

```python
success = df[df["Status"] == "SUCCESS"]
print(success)
```

```python
sorted_df = df.sort_values(
    by="Amount (INR)",
    ascending=False
)
print(sorted_df)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
top10 = sorted_df.head(10)
plt.bar(top10["Sender Name"], top10["Amount (INR)"])
plt.title("Top 10 Transaction Amounts")
plt.xlabel("Sender Name")
plt.ylabel("Amount (INR)")
plt.xticks(rotation=45)
plt.show()
```

```python
print("Mean =", df["Amount (INR)"].mean())
print("Median =", df["Amount (INR)"].median())
print("Mode =")
print(df["Amount (INR)"].mode())
print("Standard Deviation =", df["Amount (INR)"].std())
```

```python
print(df.describe())
```

```python
status_avg = df.groupby(
    "Status"
)["Amount (INR)"].mean()

print(status_avg)
```

```python
status_avg.plot(kind="bar")

plt.title("Average Transaction Amount by Status")

plt.xlabel("Status")

plt.ylabel("Average Amount (INR)")

plt.show()
```
